```python
import gradio as gr
import cv2

def object_detection(image):
    """
    This function takes an image as input,
    performs object detection using the trained model,
    and returns the detected image.
    """

    # Model prediction
    results = model.predict(image)

    # Plot detected image
    detected_image = results[0].plot()

    # Convert BGR to RGB (Gradio needs RGB)
    detected_image = cv2.cvtColor(detected_image, cv2.COLOR_BGR2RGB)

    return detected_image


# Gradio Interface
app = gr.Interface(
    fn=object_detection,
    inputs=gr.Image(type="numpy", label="Upload Image"),
    outputs=gr.Image(type="numpy", label="Detected Objects"),
    title="🧠 Object Detection System",
    description="Upload an image to detect objects using a trained AI model.",
    allow_flagging="never"
)

app.launch(share=True)
```
